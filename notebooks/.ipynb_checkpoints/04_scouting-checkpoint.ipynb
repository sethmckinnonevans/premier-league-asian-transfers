```python
import sys
import importlib
from pathlib import Path
import pandas as pd

sys.path.insert(0, str(Path.cwd().parent / "src"))

import player_plotting
importlib.reload(player_plotting)
plot_players = player_plotting.plot_players

player_data = pd.read_csv(Path.cwd().parent / "data/cleaned/scouting_percentiles.csv")
kit_colours_df = pd.read_csv(Path.cwd().parent /"data/cleaned/kit_hexcodes.csv")

kit_colours = dict(zip(kit_colours_df["team id"], kit_colours_df["hexcode"]))
```
