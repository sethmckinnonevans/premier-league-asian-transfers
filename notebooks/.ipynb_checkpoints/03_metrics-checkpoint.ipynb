### Metrics
For the scouting section of this analysis, we need to create metrics that are appropriate for plotting in order to identify stand out players. We will primarily use per 90 metrics since this gives a better representation of player's stats compared to the raw numbers, but will mean we have to enforce a minimum minutes requirement to eliminate extreme numbers due to low minutes played. We will also calculate z-scores across positional groups within a league as further analysis.


```python
# 1. Import the player data
import pandas as pd
from pathlib import Path

# Define the project folder
project_folder = Path.cwd().parent

# Import the 2026 transfers
player_data = pd.read_csv(project_folder / "data/cleaned/player_data_combined.csv")
```

```python
print(player_data.columns.tolist())
```

```python
# 2. Calculate per 90 stats
per_90_cols = ['accurateChippedPasses', 
               'accurateCrosses', 
               'accurateFinalThirdPasses', 
               'accurateLongBalls', 
               'accurateOppositionHalfPasses', 
               'accurateOwnHalfPasses', 
               'accuratePasses', 
               'aerialDuelsWon', 
               'aerialLost', 
               'assists', 
               'attemptPenaltyMiss', 
               'attemptPenaltyPost', 
               'attemptPenaltyTarget', 
               'ballRecovery', 
               'bigChancesCreated', 
               'bigChancesMissed', 
               'blockedShots', 
               'clearances', 
               'crossesNotClaimed', 
               'dispossessed', 
               'dribbledPast', 
               'duelLost', 
               'expectedAssists', 
               'expectedGoals', 
               'fouls', 
               'freeKickGoal', 
               'goalKicks', 
               'goals', 
               'goalsConceded', 
               'goalsConcededInsideTheBox', 
               'goalsConcededOutsideTheBox', 
               'goalsFromInsideTheBox', 
               'goalsFromOutsideTheBox', 
               'groundDuelsWon', 
               'headedGoals', 
               'highClaims', 
               'hitWoodwork', 
               'inaccuratePasses', 
               'interceptions', 
               'keyPasses', 
               'leftFootGoals', 
               'offsides', 
               'outfielderBlocks', 
               'ownGoals', 
               'passToAssist', 
               'penaltiesTaken', 
               'penaltyConceded', 
               'penaltyFaced', 
               'penaltyGoals', 
               'penaltySave', 
               'penaltyWon', 
               'possessionLost', 
               'possessionWonAttThird', 
               'punches', 
               'rightFootGoals', 
               'runsOut', 
               'savedShotsFromInsideTheBox', 
               'savedShotsFromOutsideTheBox', 
               'saves', 
               'savesCaught', 
               'savesParried', 
               'shotFromSetPiece', 
               'shotsFromInsideTheBox', 
               'shotsFromOutsideTheBox', 
               'shotsOffTarget', 
               'shotsOnTarget', 
               'successfulDribbles', 
               'successfulRunsOut', 
               'tackles', 
               'tacklesWon', 
               'totalAttemptAssist', 
               'totalChippedPasses', 
               'totalCross', 
               'totalLongBalls', 
               'totalOppositionHalfPasses', 
               'totalOwnHalfPasses', 
               'totalPasses', 
               'totalShots', 
               'touches', 
               'wasFouled']

minutes = player_data["minutesPlayed"]

for metric in per_90_cols:
        player_data[f"{metric}_per90"] = ((player_data[metric] / player_data["minutesPlayed"]) * 90).where(minutes > 0)
```

```python
# 3. Minimum minutes filtering
minutes_summary = (
    player_data.groupby(["season","league"])["minutesPlayed"]
    .agg(
        players="count",
        median_minutes="median",
        mean_minutes="mean",
        over_450=lambda s: (s >= 450).sum(),
        over_900=lambda s: (s >= 900).sum()
    )
)

print(minutes_summary)

# Minimum minutes by league and season
minutes_thresholds = {
    ("J1 League", 2025): 900,
    ("K League 1", 2025): 700,
    ("J1 League", 2026): 450,
    ("K League 1", 2026): 700,
    ("J1 League", "2026/27"): 225
}

player_data["season"] = player_data["season"].apply(
    lambda x: int(x) if str(x).isdigit() else str(x)
)

# Determine eligibility
player_data["minutes_threshold"] = [minutes_thresholds.get((league,season), float("nan"))
                                   for league, season in zip(
                                       player_data["league"],
                                       player_data["season"])]
player_data["eligible"] = (player_data["minutesPlayed"].ge(player_data["minutes_threshold"])
                           & player_data["minutesPlayed"].notna()
                           & player_data["minutes_threshold"].notna())


eligibility_summary = (
    player_data.groupby(["league", "season"])["eligible"]
    .agg(
        total_players="size",
        eligible_players="sum",
        eligibility_rate="mean"
    )
    .reset_index()
)

eligibility_summary["eligibility_rate"] = (
    eligibility_summary["eligibility_rate"] * 100
).round(1)

print(eligibility_summary)
```

```python
# 4. Define positional groups for scouting
position_groups = {
    "DC": "Centreback",
    "DL": "Fullback",
    "DR": "Fullback",
    "DM": "Midfielder",
    "MC": "Midfielder",
    "AM": "Midfielder",
    "ML": "Winger",
    "MR": "Winger",
    "LW": "Winger",
    "RW": "Winger",
    "ST": "Striker",
    "GK": "Goalkeeper"
}

player_data["scoutingPosition"] = (
    player_data["primaryPosition"].map(position_groups)
)

print(
    player_data.groupby(
        ["league", "season", "scoutingPosition"]
    )["eligible"]
    .sum()
    .unstack(fill_value=0)
)
```

```python
# 5. Define the cols for percentiles for positions
# extra metrics
player_data["penaltySavePercentage"] = (player_data["penaltySave"] / player_data["penaltyFaced"]).where(player_data["scoutingPosition"].eq("Goalkeeper") 
                                                                                                        & player_data["penaltyFaced"].gt(0))
player_data["savePercentage"] = player_data["saves"] / (player_data["saves"] + player_data["goalsConceded"]).where(player_data["scoutingPosition"].eq("Goalkeeper"))

# Outfielder metrics
outfielders_cols = ["accurateCrosses_per90",
                    "accurateCrossesPercentage",
                    "accurateFinalThirdPasses_per90",
                    "accurateLongBalls_per90",
                    "accurateLongBallsPercentage",
                    "accurateOppositionHalfPasses_per90",
                    "accurateOwnHalfPasses_per90",
                    "accuratePasses_per90",
                    "accuratePassesPercentage",
                    "aerialDuelsWon_per90",
                    "aerialDuelsWonPercentage",
                    "assists_per90",
                    "ballRecovery_per90",
                    "bigChancesCreated_per90",
                    "bigChancesMissed_per90",
                    "blockedShots_per90",
                    "clearances_per90",
                    "dispossessed_per90",
                    "dribbledPast_per90",
                    "duelLost_per90",
                    "errorLeadToGoal",
                    "errorLeadToShot",
                    "expectedAssists_per90",
                    "expectedGoals_per90",
                    "fouls_per90",
                    "goalConversionPercentage",
                    "goals_per90",
                    "groundDuelsWon_per90",
                    "groundDuelsWonPercentage",
                    "interceptions_per90",
                    "keyPasses_per90",
                    "minutesPlayed",
                    "passToAssist_per90",
                    "possessionWonAttThird_per90",
                    "shotsOnTarget_per90",
                    "successfulDribbles_per90",
                    "successfulDribblesPercentage",
                    "tacklesWon_per90",
                    "tacklesWonPercentage",
                    "totalShots_per90",
                    "touches_per90", 
                    "wasFouled_per90"]

# GK cols
gk_cols = ["accurateLongBalls_per90",
           "accurateLongBallsPercentage",
           "accuratePasses_per90",
           "accuratePassesPercentage",
           "cleanSheet",
           "goalsConceded_per90",
           "goalsConcededInsideTheBox_per90",
           "goalsConcededOutsideTheBox_per90",
           "highClaims_per90",
           "minutesPlayed",
           "penaltySave",
           "punches_per90",
           "savedShotsFromInsideTheBox_per90",
           "savedShotsFromOutsideTheBox_per90",
           "saves_per90",
           "successfulRunsOut_per90",
           "penaltySavePercentage",
           "savePercentage"]

# Map the metrics to positions
metrics_by_position = {
    "Centreback": outfielders_cols,
    "Fullback": outfielders_cols,
    "Midfielder": outfielders_cols,
    "Winger": outfielders_cols,
    "Striker": outfielders_cols,
    "Goalkeeper": gk_cols
}
```

```python
# 6. Calculate percentiles
# Create an empty dictionary to store percentile Series
percentile_data = {}

for position, metrics in metrics_by_position.items():

    mask = (
        player_data["scoutingPosition"].eq(position)
        & player_data["eligible"]
    )

    for metric in metrics:
        percentile_col = f"{metric}_pctl"

        # Initialise the percentile Series
        if percentile_col not in percentile_data:
            percentile_data[percentile_col] = pd.Series(
                float("nan"),
                index=player_data.index
            )

        # Calculate percentile among eligible players in this position
        percentile_data[percentile_col].loc[mask] = (
            player_data.loc[mask, metric].rank(pct=True) * 100
        )

# Add all percentile columns to the DataFrame in one operation
percentile_df = pd.DataFrame(percentile_data, index=player_data.index)

# Replace existing percentile columns if rerunning this cell
player_data = player_data.drop(
    columns=percentile_df.columns.intersection(player_data.columns)
)

player_data = pd.concat([player_data, percentile_df], axis=1)
```

```python
# 7. Export the data
player_data.to_csv(project_folder / "data/cleaned/scouting_percentiles.csv", index = False)
```
